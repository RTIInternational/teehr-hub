## Summary
#### This is just a cleanup notebook to drop tables from the Iceberg warehouse that are no longer relevant.

```python
import teehr
```

```python
from teehr.evaluation.spark_session_utils import create_spark_session

spark = create_spark_session()

ev = teehr.RemoteReadWriteEvaluation(spark=spark)
```

```python
df = ev.list_tables().sort_values(by="name")
```

```python
df.to_csv("teehr-warehouse-tables-2026.09.28.csv", index=False)
```

```python
tables_to_remove = [
    "configuration_completeness",
    "forecast_metrics_by_location",
    "grid_pixel_fractional_coverage",
    "joined_forecast_timeseries",
    "joined_simulation_timeseries",
    "nwmd_metrics_by_location_v3_auto",
    "locations_with_attributes"
]
```

```python
ev.spark.sql("SHOW namespaces").show()
```

```python
for table_name in tables_to_remove:
    print(f"Dropping table: {table_name}")
    spark.sql(f"DROP TABLE iceberg.teehr.{table_name} PURGE")
```

```python
ev.list_tables().sort_values(by="name")
```

```python
ev.spark.stop()
```
